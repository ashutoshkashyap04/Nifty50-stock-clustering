```python
all_data = {}

for ticker in nifty50:
    
    try:
        print(f"Downloading {ticker}")
        
        df = yf.download(
            ticker,
            start= "2021-01-01",
            end= "2025-12-31",
            auto_adjust = True
        )
        
        if not df.empty:
            all_data[ticker] = df
        else:
            print(f"No data for {ticker}")
            
    except Exception as e:
        print(f"failed: {ticker} -> {e}")
```

```python
for ticker, df in all_data.items():
    df.to_csv(f"data/raw/{ticker}.csv")
```

# EDA and Data Cleaning

```python
# dataset overview 

print("Total Stocks", len(all_data))

for ticker, df in all_data.items():
    print(f"{ticker}: Shape = {df.shape}")
    print(df.columns)
    print("-" * 50)
```

```python
# Checking datatypes

for ticker, df in all_data.items():
    print(f"\n{ticker}")
    print(df.dtypes)
    break 
```

```python
# Checking null and duplicates

for ticker, df in all_data.items():
    if df.isnull().sum().sum() > 0:
        print('\n{ticker}')
        print(df.isnull().sum())
    
    if df.duplicated().sum() > 0:
        duplicates = df.duplicated().sum()
        print(f'Duplicates in {ticker}: {duplicates}')
        df = df.drop_duplicates(inplace= True)
        print(f'Duplicates in {ticker}: {df.duplicated().sum()}')
        
```

```python
# date range

date_ranges = {}

for ticker, df in all_data.items():
    date_ranges[ticker] = (df.index.min(), df.index.max())

print(set(date_ranges.values()))
```

```python
for ticker, df in all_data.items():
    if df.index.min() != pd.Timestamp('2021-01-01'):
        print(f'{ticker} : {df.index.min().date()} to {df.index.max()}')
```

```python
#what are the daily returns 

returns_data = {}

for ticker, df in all_data.items():
    returns_data[ticker] = (df['Close'].squeeze().pct_change()) * 100
    
returns_df = pd.DataFrame(returns_data)

returns_df.head()
```

```python
returns_df.describe()
```

```python
# average and median returns 
mean_returns = returns_df.mean()

plt.figure(figsize=(12,6))
mean_returns.plot(kind="bar")
plt.title("Average Daily Return of Nifty 50 Stocks")
plt.xlabel("Stocks")
plt.ylabel("Mean Daily Return")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
median_returns = returns_df.median()

plt.figure(figsize=(12,6))
median_returns.plot(kind="bar")
plt.title("Median Return of Nifty 50 Stocks")
plt.xlabel("Stocks")
plt.ylabel("Median Return")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# most volatile stocks

volatility = returns_df.std().sort_values()

plt.figure(figsize=(12,6))
volatility.plot(kind="bar")
plt.title("Volatility of Nifty 50 Stocks")
plt.xlabel("Stocks")
plt.ylabel("Standard Deviation of Daily Returns")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
# Correlation of stocks

correlation_matrix = returns_df.corr()


plt.figure(figsize= (12,10))

sns.heatmap(correlation_matrix, cmap= 'coolwarm')
plt.title("Nifty 50 correlation heatmap")
plt.show()
```
