```python
all_data = {}

for ticker in nifty50:
    
    try:
        print(f"Downloading {ticker}")
        
        df = yf.download(
            ticker,
            start= "2021-01-01",
            end= "2025-12-31",
            auto_adjust = True
        )
        
        if not df.empty:
            all_data[ticker] = df
        else:
            print(f"No data for {ticker}")
            
    except Exception as e:
        print(f"failed: {ticker} -> {e}")
```

```python
for ticker, df in all_data.items():
    df.to_csv(f"data/raw/{ticker}.csv")
```

# Feature Engineering 

```python
#annualized returns

annual_return = mean_returns * 252
# average number of trading days in a year = 252
```

```python
# annualized volatility

annual_volatility = volatility * np.sqrt(252)
# sqrt(252) because volatility(standard deviation) scales with the square root of time
```

```python
# Sharpe ratio- measures how much returns stock generated per unit risk

risk_free_rate = 0.07 
# risk free rate is the return that can be earned without taking investment risk

sharpe_ratio  = (annual_return - risk_free_rate) / annual_volatility
```

```python
# Max drawdown - biggest percentage loss from a peak before the stock recovered

max_drawdown = {}

for ticker in returns_df.columns:
    cumulative_returns  = (1 + returns_df[ticker]).cumprod()
    running_max = cumulative_returns.cummax()
    drawdown = (cumulative_returns - running_max) / running_max
    max_drawdown[ticker] = drawdown.min()
    
max_drawdown = pd.Series(max_drawdown)
```

```python
features = pd.DataFrame({
    "Annual Return": annual_return,
    "Annual Volatility": annual_volatility,
    "Sharpe Ratio": sharpe_ratio,
    "Max Drawdown": max_drawdown
})

features.head()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

scaled_features = scaler.fit_transform(features)
```

```python
scaled_features = pd.DataFrame(
    scaled_features,
    columns= features.columns,
    index= features.index
)

scaled_features.head()
```

```python
scaled_features.describe()
```

# Model Building

```python
from sklearn.cluster import KMeans
```

```python
wcss = []    #within cluster sum of squares

for k in range(1, 11):
    kmeans = KMeans(n_clusters= k, random_state= 42, n_init= 10)
    kmeans.fit(scaled_features)
    wcss.append(kmeans.inertia_)
```

```python
#Plotting

plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), wcss, marker='o')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS (Inertia)")
plt.title("Elbow Method")
plt.show()
```

```python
# Train KMeans

kmeans = KMeans(
    n_clusters= 4,
    random_state= 42,
    n_init= 10
)

kmeans.fit(scaled_features)
```

```python
# Getting the cluster labels

clusters = kmeans.labels_
clusters
```

```python
# Add the cluster labels in dataframe

features['Cluster'] = clusters
features
```

```python
features['Cluster'].value_counts()
```

```python
cluster_summary = features.groupby('Cluster').mean()
cluster_summary
```

```python
# Visualisation using PCA

from sklearn.decomposition import PCA

pca = PCA(n_components= 2)

pca_features = pca.fit_transform(scaled_features)
pca_features
```

```python
pca_df = pd.DataFrame(
    pca_features,
    columns= ['PC1', 'PC2']
)

pca_df['Cluster'] = clusters

pca_df
```

```python
plt.figure(figsize=(8,6))

for cluster in sorted(pca_df["Cluster"].unique()):
    subset = pca_df[pca_df["Cluster"] == cluster]

    plt.scatter(
        subset["PC1"],
        subset["PC2"],
        label=f"Cluster {cluster}"
    )

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("K-Means Clustering of Nifty 50 Stocks")
plt.legend()
plt.grid(True)
plt.show()
```
